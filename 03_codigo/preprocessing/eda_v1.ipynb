# ANÁLISIS DE LA BASE DE DATOS CIC-IDS-2017

```python
import pandas as pd
import glob
import os

directorio_actual = os.getcwd()
nombre_dataset = "CIC17"
directorio_dataset = os.path.abspath(
    os.path.join(directorio_actual, "../../02_datasets/raw", nombre_dataset)
)

archivos = glob.glob(directorio_dataset + "/*.csv")

df_list = []

for archivo in archivos:
    df_temp = pd.read_csv(archivo)
    df_list.append(df_temp)

df = pd.concat(df_list, ignore_index=True)
```

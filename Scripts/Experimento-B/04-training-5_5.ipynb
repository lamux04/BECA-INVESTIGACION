```python
import re
from os import listdir
import pandas as pd
from simplet5 import SimpleT5
```

```python
experiment_code = 'B'
version = 1
fold = 5
k = 5
experiment_name = f'experimento-{experiment_code}-overfitting'
model_name = 't5-small'
dataset_path = './dataset'
model_path = '../models/' + model_name
```

```python
dataset_basename = '{}-03-preparing-dataset-pt-{}'.format(experiment_code, fold)
train_path = '{}/{}-train.csv'.format(dataset_path, dataset_basename)
val_path = '{}/{}-val.csv'.format(dataset_path, dataset_basename)
print('train path: ' + train_path)
print('val path: ' + val_path)

basename = '04-{}-v{}-{}_{}-{}-{}'.format(experiment_code, version, fold, k, experiment_name, model_name)
output_path = '{}-1-training-outputs'.format(basename)
print('output path: ' + output_path)
```

```python
df_train = pd.read_csv(train_path)
df_train['source_text'] = df_train['source_text'].astype(str)
df_train['target_text'] = df_train['target_text'].astype(str)
```

```python
df_train.head()
```

```python
df_val = pd.read_csv(val_path)
df_val['source_text'] = df_val['source_text'].astype(str)
df_val['target_text'] = df_val['target_text'].astype(str)
```

```python
model_train_data = SimpleT5()
model_train_data.from_pretrained("t5", model_path)
```

```python
model_train_data.train(train_df=df_train, eval_df=df_val,
            source_max_token_len=150, 
            target_max_token_len=3,
            batch_size=16,
            precision=32,
            max_epochs=10, 
            outputdir=output_path,
            use_gpu=True,
            dataloader_num_workers=24)
```

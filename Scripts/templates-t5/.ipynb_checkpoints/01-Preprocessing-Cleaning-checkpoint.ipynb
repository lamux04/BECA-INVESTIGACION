```python
import pandas as pd
import numpy as np
import re
```

```python
import sys
sys.path.insert(0, '../src')

from preprocessing import (
    get_file_list,
    open_and_merge_dataset, 
    datetime_to_timestamp,
    sanitize_column_names, 
    remove_columns_std_dev_equals_zero, 
    remove_columns_high_correlation, 
    replace_inf_by_nan, 
    remove_missing_values, 
    remove_duplicate_rows,
    save_dataframe,
    save_dataframe_split,
    print_shape,
    print_object_columns,
    print_existing_columns,
    remove_columns,
    print_column_types,
    get_columns_std_dev_equals,
    get_columns_high_correlation
)
```

```python
path = '../dataset/'
```

```python
df = open_and_merge_dataset(path)
```

```python
df_clean = (df.
        pipe(print_shape).
        pipe(print_object_columns).
        pipe(datetime_to_timestamp, 'Timestamp').
        pipe(sanitize_column_names).
        pipe(print_object_columns).
        pipe(print_existing_columns, ['FLOW_ID', 'SRC_IP', 'SRC_PORT', 'DST_IP']).
        pipe(remove_columns, ['FLOW_ID', 'SRC_IP', 'SRC_PORT', 'DST_IP']).
        pipe(print_shape).
        pipe(print_object_columns).
        pipe(print_column_types).
        pipe(remove_columns_std_dev_equals_zero).
        pipe(print_shape).
        pipe(remove_columns_high_correlation).
        pipe(print_shape).
        pipe(replace_inf_by_nan).
        pipe(print_shape).
        pipe(remove_missing_values).
        pipe(print_shape)
    )
```

```python
path_carpeta = "../dataset/CSE-2018"
path_database = "05-01-06-prepared-dataset-no-dup-L1-all-labels.csv"
name_target = "LABEL"
```

```python
import pandas as pd
data = pd.read_csv(f'{path_carpeta}/{path_database}')
y = data[name_target]

print("NÚMERO DE PATRONES EN CADA CLASE")
n_apariciones = y.value_counts()
import matplotlib.pyplot as plt
plt.bar(n_apariciones.index, n_apariciones.values)
print(n_apariciones)
```

```python
# Comprobar filas duplicadas
print("NÚMERO DE FILAS DUPLICADAS POR CLASE")
duplicados = data[data.duplicated()]
if duplicados.empty:
    print("No hay filas duplicadas en el dataset.")
else:
    # Asegurarse de que la columna existe
    if name_target in data.columns:
        n_duplicados = duplicados[name_target].value_counts()
        print(n_duplicados)
        # Si quieres graficar:
        # plt.bar(n_duplicados.index, n_duplicados.values)
        # plt.show()
    else:
        print(f"La columna '{name_target}' no existe. Las columnas disponibles son:")
        print(data.columns.tolist())
```

# ANÁLISIS DE LA BASE DE DATOS CSE-CIC-2018

```python
import pandas as pd
import glob
import os

directorio_actual = os.getcwd()
nombre_dataset = "CIC18"
directorio_dataset = os.path.abspath(
    os.path.join(directorio_actual, "../../02_datasets/raw", nombre_dataset)
)

archivos = glob.glob(directorio_dataset + "/*.csv")
archivos.sort()

df_list = []

for archivo in archivos:
    print(f"Leyendo: {archivo}")
    df_temp = pd.read_csv(archivo, low_memory=False, on_bad_lines='warn')
    df_list.append(df_temp)

df = pd.concat(df_list, ignore_index=True)
```

```python
################## COLUMNAS CON ALTA CORRELACIÓN #####################
import numpy as np
import pandas as pd

df_datos = df.drop(columns="Label")

# Solo columnas numéricas
df_num = df_datos.select_dtypes(include=np.number)

corr_matrix = df_num.corr().abs()

high_corr = (
    corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .sort_values(ascending=False)
)

high_corr[high_corr > 0.95].shape
```

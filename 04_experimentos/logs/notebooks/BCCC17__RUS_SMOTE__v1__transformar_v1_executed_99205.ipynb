```python
nombre_train = "BCCC17__RUS_SMOTE__v1__train.csv"
nombre_test = "BCCC17__RUS_SMOTE__v1__test.csv"
ruta_base_dataset = "../../../02_datasets/processed/BCCC17__RUS_SMOTE__v1"
nombre_train_generado = "pd_BCCC17__RUS_SMOTE__v1__train.csv"
nombre_test_generado = "pd_BCCC17__RUS_SMOTE__v1__test.csv"
nombre_test_nuevo = nombre_test.replace(".csv", "_nuevo.csv")
nombre_test_nuevo_generado = nombre_test_generado.replace(".csv", "_nuevo.csv")
ruta_base_dataset_generado = "../../../02_datasets/processed/pd_BCCC17__RUS_SMOTE__v1"
```

```python
df_train = cargar_dataset(nombre_train, ruta_base_dataset)
df_test = cargar_dataset(nombre_test, ruta_base_dataset)
df_test_nuevo = cargar_dataset(nombre_test_nuevo, ruta_base_dataset)
```

```python
particiones = dividir_en_k_particiones_stratified(df_train, "LABEL", 5)
```

```python
k = 1
for df_part_train, df_part_val in particiones:
    df_string_train = dataframe_a_source_target(df_part_train)
    df_string_val = dataframe_a_source_target(df_part_val)
    guardar_dataset_csv(df_string_train, nombre_train_generado.replace(".csv", f"__{k}.csv"), ruta_base_dataset_generado)
    guardar_dataset_csv(df_string_val, nombre_train_generado.replace("train.csv", f"val__{k}.csv"), ruta_base_dataset_generado)
    k += 1
```

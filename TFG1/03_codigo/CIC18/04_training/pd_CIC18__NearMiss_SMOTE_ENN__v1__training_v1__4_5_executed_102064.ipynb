```python
from pathlib import Path
import re
from os import listdir
import pandas as pd
```

```python
k = 4
BASE_DIR = Path("../../../").resolve()
ruta_base_dataset = str(BASE_DIR / "02_datasets" / "processed" / "pd_CIC18__NearMiss_SMOTE_ENN__v1")
model_path = str(BASE_DIR / "04_experimentos" / "modelos" / "t5-small")
output_path = str(BASE_DIR / "04_experimentos" / "modelos" / "pd_CIC18__NearMiss_SMOTE_ENN__v1__outputs" / f"pd_CIC18__NearMiss_SMOTE_ENN__v1__outputs__{k}_5")

nombre_train = f"pd_CIC18__NearMiss_SMOTE_ENN__v1__train__{k}.csv"
nombre_val = f"pd_CIC18__NearMiss_SMOTE_ENN__v1__val__{k}.csv"
```

```python
import sys
sys.path.append("../../utils/")

from utils import *
```

```python
df_train = cargar_dataset(nombre_train, ruta_base_dataset)
df_val = cargar_dataset(nombre_val, ruta_base_dataset)

df_train['source_text'] = df_train['source_text'].astype(str)
df_train['target_text'] = df_train['target_text'].astype(str)
df_val['source_text'] = df_val['source_text'].astype(str)
df_val['target_text'] = df_val['target_text'].astype(str)
```

```python
model_train_data = SimpleT5Wrapper()
model_train_data.from_pretrained("t5", model_path, use_gpu=True)
```

```python
model_train_data.train(
    train_df=df_train,
    eval_df=df_val,
    source_max_token_len=150,
    target_max_token_len=3,
    batch_size=16,
    precision=32,
    max_epochs=10,
    outputdir=output_path,
    use_gpu=True,
    dataloader_num_workers=16
)
```
